### Part 1: Practice resampling & chart building with Pandas and Matplotlib

```python
# For this assignment, you'll be working with NYC payroll data again. 
# Import it from wherever you stored the data

payroll_data = pd.read_csv('../data/Citywide_Payroll_Data__Fiscal_Year__20240616.csv')
payroll_data
```

Using this data, create an ordered, horizontal bar chart of the top 10 biggest agencies in NYC. 

```python
# Your code here.

# Ok first I need to figure out how many people work in each agency.

payroll_data_size = payroll_data["Agency Name"].value_counts().head(10)


# Now I need to plot it

payroll_data_size.plot(kind="barh", x="NUMBER OF EMPLOYEES", y="AGENCY", color="blue")
```

Now, create an ordered, horizontal bar chart of the top ten agencies with the highest median base salary.

```python
# Your code here.

# First I need to figure out what the top ten agencies with the highest median base salary are. 

# payroll_data.dtypes

payroll_data_salary = payroll_data.groupby(["Agency Name"])

payroll_median_salary = payroll_data_salary["Base Salary"].median().reset_index().sort_values(by="Base Salary", ascending=False).head(10)

payroll_median_salary.plot(kind="barh", x="Agency Name", y="Base Salary")
```

Let's return to our police data to practice resampling. 

```python
# Remember to change the file path to match where your data is stored. 

police_data = pd.read_csv("../data/fatal-police-shootings-data.csv", parse_dates=["date"])
police_data
```

```python
police_data.head()
```

Remember in class, we learned that resampling requires several basic steps:

1. Make sure that Pandas is reading in your date column as a date. If it's not, convert it to datetime.
2. Create a new dataframe with your date column as your index
3. Resample the data over the time interval of your choosing.

With this in mind, resample the data over a monthly period and plot it in a line chart.

```python
# Your code here.

# Confirm data type
# police_data.dtypes

# Ok, now I need to reindex it by date.
police_date_index = police_data.set_index("date")

# Now to resample...

police_time_month = police_date_index.resample('ME')["id"].count()


police_time_month
```

Now create an ordered bar chart of total police killings by year. Put the two charts side by side using the strategy we reviewed in class.

```python
# Your code here

# Ok now I just have to do the last step of resampling, but this time by year. 

police_time_year = police_date_index.resample('YE')["id"].count()

police_time_year
```

```python
# New kernal for the plotting:

fig, ax = plt.subplots(1,2)

fig.set_size_inches(12,4)

ax[0].plot(police_time_month, color = "red")
ax[1].plot(police_time_year, color = "blue")
```
